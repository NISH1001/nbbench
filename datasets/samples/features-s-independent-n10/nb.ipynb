```python
import os
import sys
import tempfile
import time

P = {'rows': 10500, 'cols': 10}
DIR = os.environ.get("NBBENCH_DIR") or tempfile.mkdtemp(prefix="nbbench-")
HOST = "127.0.0.1"
PORT = int(os.environ.get("NBBENCH_PORT", "0"))


def finish(spans, digests):
    import hashlib
    import json

    checksum = hashlib.blake2b("".join(digests).encode()).hexdigest()[:16]
    result = {"prefix": sys.prefix, "spans": spans, "digests": digests, "checksum": checksum, "done": time.time()}
    with open(os.path.join(DIR, "result.json"), "w") as out:
        json.dump(result, out)
    print("checksum", checksum)


import numpy as np
import pandas as pd

RAW = pd.DataFrame(
    np.random.default_rng(0).random((P["rows"], P["cols"])),
    columns=[f"f{k}" for k in range(P["cols"])],
)


def initial():
    return np.ones(P["cols"])


def work(x, i):
    weighted = RAW * x
    standard = (weighted - weighted.mean()) / weighted.std()
    logs = np.log1p(weighted).add_suffix("_log")
    pairs = pd.DataFrame({
        f"f{k}_x_f{k + 1}": standard[f"f{k}"] * standard[f"f{k + 1}"] for k in range(P["cols"] - 1)
    })
    table = pd.concat([standard, logs, pairs], axis=1)
    correlation = table.corr().to_numpy()[: P["cols"], P["cols"]:]
    scores = np.abs(correlation).mean(axis=1)
    return scores / scores.max()


def combine(xs):
    return np.mean(xs, axis=0)


def digest(x):
    return f"{float(x.sum()):.6e}"


x0 = initial()
print('setup', 'features')
```

```python
t0_1 = time.time()
x1 = work(x0, 1)
span1 = (t0_1, time.time())
d1 = digest(x1)
print(1, d1)
```
