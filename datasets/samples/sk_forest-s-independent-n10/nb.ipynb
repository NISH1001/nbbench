```python
import os
import sys
import tempfile
import time

P = {'rows': 1000, 'trees': 9}
DIR = os.environ.get("NBBENCH_DIR") or tempfile.mkdtemp(prefix="nbbench-")
HOST = "127.0.0.1"
PORT = int(os.environ.get("NBBENCH_PORT", "0"))


def finish(spans, digests):
    import hashlib
    import json

    checksum = hashlib.blake2b("".join(digests).encode()).hexdigest()[:16]
    result = {"prefix": sys.prefix, "spans": spans, "digests": digests, "checksum": checksum, "done": time.time()}
    with open(os.path.join(DIR, "result.json"), "w") as out:
        json.dump(result, out)
    print("checksum", checksum)


import numpy as np
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier

FEATURES, LABELS = make_classification(
    n_samples=P["rows"], n_features=16, n_informative=8, random_state=0
)


def initial():
    return np.ones(16)


def work(x, i):
    model = RandomForestClassifier(n_estimators=P["trees"], max_depth=8, random_state=i, n_jobs=1)
    model.fit(FEATURES * x, LABELS)
    importance = model.feature_importances_
    return 0.5 + importance / importance.max()


def combine(xs):
    return np.mean(xs, axis=0)


def digest(x):
    return f"{float(x.sum()):.4e}"


x0 = initial()
print('setup', 'sk_forest')
```

```python
t0_1 = time.time()
x1 = work(x0, 1)
span1 = (t0_1, time.time())
d1 = digest(x1)
print(1, d1)
```
